```python

import cv2

# Step 1: Import libraries and initialize webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Could not open webcam")
    exit()

print("Webcam started successfully!")

# Step 2: Capture frames continuously
while True:
    ret, frame = cap.read()

    if not ret:
        print("Error: Cannot capture frame")
        break

    # Step 3: Save a frame as JPG image
    cv2.imwrite("webcam_capture.jpg", frame)

    # Step 4: Display live video stream
    cv2.imshow("Live Webcam", frame)

    # Step 5: Resize the frame
    resized_frame = cv2.resize(frame, (640, 480))
    cv2.imshow("Resized Frame", resized_frame)

    # Rotate the frame by 90 degrees clockwise
    rotated_frame = cv2.rotate(
        frame, cv2.ROTATE_90_CLOCKWISE
    )
    cv2.imshow("Rotated Frame", rotated_frame)

    # Press Q to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release webcam and close windows
cap.release()
cv2.destroyAllWindows()

print("Webcam stopped successfully!")
```
